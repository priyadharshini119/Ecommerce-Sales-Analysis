```python
import pandas as pd
```

```python
df = pd.read_csv("cleaned_data.csv")
```

```python
import sqlite3
```

```python
conn = sqlite3.connect("analysis.db")
```

```python
df.to_sql("sales",conn,if_exists="replace",index=False)
```

```python
query = "SELECT * FROM sales LIMIT 20;"
```

```python
result = pd.read_sql(query, conn)
```

```python
query = "SELECT Country, SUM(Quantity * Unitprice) AS total_revenue FROM sales Group by country order by total_Revenue DESc Limit 10;"
```

```python
result = pd.read_sql(query, conn)
```

```python
result
```

```python
query = " SELECT strftime('%H', InvoiceDate) AS hour,COUNT(*) AS purchases FROM sales GROUP BY hour ORDER BY purchases DESC;"
```

```python
result
```

```python
query = " SELECT CustomerID,COUNT(DISTINCT InvoiceNo) AS orders FROM sales WHERE CustomerID IS NOT NULL GROUP BY CustomerID;"
```

```python
result = pd.read_sql(query, conn)
```

```python
result
```

```python
query = "SELECT AVG(basket_size) AS average_basket_size FROM (SELECT InvoiceNo, SUM(Quantity) AS basket_size FROM sales GROUP BY InvoiceNo);"
```

```python
result = pd.read_sql(query, conn)
```

```python
result
```

```python
query = "SELECT strftime('%Y-%m', InvoiceDate) AS month,SUM(Quantity * UnitPrice) AS revenue FROM sales GROUP BY month ORDER BY month;"
```

```python
result = pd.read_sql(query, conn)
```
